```python
password = input("Enter your password: ")

special_chars = ["@", "#", "$", "%", "&"]

rules = [
    len(password) >= 8,                     # length check
    any(c.isdigit() for c in password),     # has digit
    any(c.isupper() for c in password),     # has uppercase
    any(c in special_chars for c in password)  # has special char
]

# conditional check
if all(rules):
    strength = "Strong"
elif rules[0] and (rules[1] or rules[2]):
    strength = "Medium"
else:
    strength = "Weak"

print("Password Strength:", strength)
```
